# Prepare offline assets

Run this on Kaggle with internet disabled. Update the dataset folder names in the first code cell.

```python
from pathlib import Path
import os
import subprocess
import sys

INPUT = Path('/kaggle/input')
WORK = Path('/kaggle/working/rsna-knee')
PACKAGE = INPUT / 'rsna-knee-code' / 'rsna-knee-abnormality-detection' / 'src'
COMPETITION = INPUT / 'rsna-knee-abnormality-detection'
CHECKPOINT = INPUT / 'dinov2-offline' / 'dinov2_vits14_pretrain.pth'

assert PACKAGE.exists(), PACKAGE
assert COMPETITION.exists(), COMPETITION
assert CHECKPOINT.is_file(), CHECKPOINT
WORK.mkdir(parents=True, exist_ok=True)
os.environ['PYTHONPATH'] = str(PACKAGE)
PYTHON = sys.executable
```

```python
subprocess.run([
    PYTHON, '-m', 'rsna_knee', 'feature-fingerprint',
    '--checkpoint', str(CHECKPOINT),
], check=True)
```

```python
subprocess.run([
    PYTHON, '-m', 'rsna_knee', 'build-protocol-manifest',
    '--series-csv', str(COMPETITION / 'train_series.csv'),
    '--output', str(WORK / 'train-protocol-manifest-v1.csv'),
], check=True)
```
